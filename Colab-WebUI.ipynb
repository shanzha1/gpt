# GPT-SoVITS WebUI

## Env Setup (Run Once Only)
## 环境配置, 只需运行一次

### 1.

```
%%writefile /content/setup.sh
set -e

cd /content

if [ ! -d "GPT-SoVITS" ]; then
    git clone https://github.com/RVC-Boss/GPT-SoVITS.git
fi

cd GPT-SoVITS

# 确保 Conda 环境存在
if ! conda env list | grep -q "GPTSoVITS"; then
    conda create -n GPTSoVITS python=3.10 -y
fi

# 使用 conda run 来确保在正确的环境下执行安装，无需关心绝对路径
conda run -n GPTSoVITS pip install --upgrade pip

# 安装兼容 12.8 的 PyTorch 12.4 版本
conda run -n GPTSoVITS pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu124

# 安装项目依赖
conda run -n GPTSoVITS pip install -r requirements.txt

# 运行模型下载
bash install.sh --device CPU --source HF --download-uvr5 || true

echo "环境配置完成。"
```
